# Customer churn: business summary

## Objective

The objective is to identify customers at higher risk of churn so that a telecom retention team can prioritize proactive interventions.

This project uses the IBM Telco Customer Churn sample dataset and demonstrates an end-to-end, reproducible machine-learning workflow: data quality checks, leakage-safe preprocessing, cross-validation, experiment tracking with MLflow and model interpretation with permutation importance and SHAP.

## Model performance

The selected model was logistic regression, chosen after stratified 5-fold cross-validation because it achieved comparable or slightly better ROC-AUC than random forest while remaining easier to interpret.

| Metric | Test result |
|---|---:|
| ROC-AUC | 0.842 |
| Accuracy | 0.806 |
| Precision | 0.657 |
| Recall | 0.559 |
| F1 score | 0.604 |

The close agreement between cross-validation ROC-AUC (0.846) and test ROC-AUC (0.842) suggests stable performance on unseen data.

## Key predictive signals

The interpretation analyses highlighted several variables consistently associated with churn predictions:

- Lower customer tenure was the strongest predictive signal of higher churn risk.
- Month-to-month contracts and fiber-optic service were frequently associated with increased predicted risk.
- Contract type, billing preferences, payment method and support-related features also contributed to the model's predictions.
- These are predictive associations, not evidence of causal relationships.

## Recommended retention actions

1. Prioritize newly acquired customers for early retention outreach, especially during their first months.
2. Design targeted offers for customers on month-to-month contracts, such as incentives to move to longer commitments.
3. Review the experience, pricing and support needs of high-risk fiber-optic customers.
4. Use predicted risk as a prioritization signal alongside business rules, customer value and campaign cost.
5. Measure the outcome of interventions with controlled experiments before deploying them at scale.

## Limitations

- The dataset is a public, fictional telecommunications sample and may not represent a real customer population.
- Model explanations identify predictive patterns, not causal effects.
- The default classification threshold may not reflect the financial trade-off between retention actions and lost customers.
- Customer behaviour, pricing and product offerings can change over time; production models require monitoring and periodic retraining.

## Next steps

- Select a decision threshold using retention campaign cost and expected customer value.
- Evaluate calibration to ensure predicted probabilities reflect observed churn rates.
- Track data quality, drift and model performance after deployment.
- Test retention interventions through controlled experiments to measure real business impact.
